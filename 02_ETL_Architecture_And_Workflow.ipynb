## 02 — ETL Architecture & Workflow

### Project Objective
Build a scalable and reliable data pipeline that transforms raw e-commerce data into clean, validated, and business-ready analytics datasets.

### Architecture
- Source Layer: Raw Olist datasets
- Bronze Layer: Raw data ingestion and storage
- Silver Layer: Data cleaning, transformation, and validation
- Gold Layer: Business analytics and KPI tables
- Consumption Layer: Dashboards and reporting

### Workflow
1. Extract source data
2. Ingest into Bronze
3. Transform Bronze into Silver
4. Perform data quality validation
5. Route invalid data to quarantine
6. Generate Gold analytics
7. Monitor pipeline execution
8. Deliver data to dashboards

### Technology Stack
- Databricks
- Apache Spark
- PySpark
- SQL
- Delta Lake
- Medallion Architecture

### Pipeline Reliability
- Data quality checks
- Reconciliation
- Logging
- Error handling
- Incremental processing
- Pipeline monitoring


Architecture শেখার প্রথম নিয়ম \
কোনো project-এর architecture বানানোর আগে এই ৫টি প্রশ্ন করবে:\
1. Source: Data কোথা থেকে আসবে?\
2. Storage: Data কোথায় store করব?\
3. Transformation: Data কীভাবে clean ও process হবে?\
4. Orchestration: কাজগুলো কোন sequence-এ চলবে?\
5. Consumption: Final data কে এবং কীভাবে ব্যবহার করবে?\
এই ৫টি প্রশ্নের উত্তর পেলেই architecture design শুরু করা যায়। এগুলো মুখস্থ করার বিষয় নয়—project-এর requirement থেকে\বের করতে হয়।

## Workflow Orchestration

### Execution Sequence
1. Bronze Data Ingestion
2. Silver Data Transformation
3. Data Quality Validation
4. Gold Analytics Generation
5. Business KPI Update

### Dependencies
- Silver depends on Bronze.
- Data Quality depends on Silver.
- Gold depends on successful Data Quality validation.
- KPI generation depends on Gold tables.

### Failure Handling
- Failed tasks should stop dependent tasks.
- Invalid records should be sent to quarantine.
- Failed tasks should support retry.
- Pipeline execution should be logged and monitored.

## Notebook Task Design

### Task 1 — Data Ingestion
**Input:** Source CSV datasets  
**Process:** Read and ingest raw source data  
**Output:** Bronze Delta tables  
**Dependency:** None

### Task 2 — Data Transformation
**Input:** Bronze tables  
**Process:** Cleaning, standardization, transformation  
**Output:** Silver tables  
**Dependency:** Task 1

### Task 3 — Data Quality Validation
**Input:** Silver tables  
**Process:** Null checks, duplicate checks, business rules, reconciliation  
**Output:** Validation results and quarantine records  
**Dependency:** Task 2

### Task 4 — Gold Analytics
**Input:** Validated Silver data  
**Process:** Business aggregations and analytical calculations  
**Output:** Gold analytics tables  
**Dependency:** Task 3

### Task 5 — KPI and Monitoring
**Input:** Gold tables and pipeline execution metadata  
**Process:** KPI update, pipeline status, execution monitoring  
**Output:** Business KPI datasets and monitoring records  
**Dependency:** Task 4
